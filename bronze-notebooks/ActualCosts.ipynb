```python
base_path = '/Workspace/Users/hnagarkar@gmail.com/data'
catalog = '`de-foundation-01`'
```

```python
from pyspark.sql.functions import col, current_timestamp



bronze_actual = (
    spark.read
    .format("csv")
    .option("header", True)
    .option("inferSchema", False)
    .load(f"{base_path}/erp/ActualCost_*.csv")
    .select(
        "*",
        col("_metadata.file_name").alias("_source_file"),
        current_timestamp().alias("_ingested_at")
    )
)

display(bronze_actual)
```

```python
(
    bronze_actual.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(
        f"{catalog}.bronze.actual_cost"
    )
)
```
