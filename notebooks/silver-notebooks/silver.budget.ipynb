```python
catalog = '`de-foundation-01`'
```

```python
from pyspark.sql.functions import (
    coalesce,
    lit,
    when,
    try_to_date,
    col
)
silver_budget = (
    spark.table(f"{catalog}.bronze.budget")
    .withColumn(
        "Month_End_Date",
        try_to_date(
            col("Month_End_Date"),
            "yyyy-MM-dd"
        )
    )
    .withColumn(
        "Budget_Amount",
        col("Budget_Amount")
        .cast("decimal(18,2)")
    )
    .withColumn(
        "Budget_Version",
        col("Budget_Version").cast("int")
    )
)

(
    silver_budget.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(
        f"{catalog}.silver.budget"
    )
)
```
