```python
catalog = '`de-foundation-01`'
```

```python
from pyspark.sql.functions import trim, to_date, col

silver_project = (
    spark.table(f"{catalog}.bronze.project")
    .withColumn(
        "Project_Code",
        trim(col("Project_Code"))
    )
    .withColumn(
        "Start_Date",
        to_date(col("Start_Date"), "yyyy-MM-dd")
    )
    .withColumn(
        "End_Date",
        to_date(col("End_Date"), "yyyy-MM-dd")
    )
    .dropDuplicates(["Project_Code"])
)

(
    silver_project.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(
        f"{catalog}.silver.project"
    )
)
```
